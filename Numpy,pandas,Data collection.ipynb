```python
import pandas as pd
df=pd.DataFrame({
    "country":["usa","usa","india","india"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,15,18,22]
})
melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metrics",
    value_name="value")
original=melted_df.pivot(
    index=["country","year"],
    columns="metrics",
    values="value")
original
```

```python
from bs4 import BeautifulSoup
with open("scraped_data/data.html","r") as f:
    html_content=f.read()
soup=BeautifulSoup(html_content,"html.parser")
all_h3=soup.find_all("h3")
all_countries=[]
for h3 in all_h3:
    name=h3.get_text(strip=True)
    population=(h3.find_next("div").select_one("span.country-population").get_text(strip=True))
    all_countries.append([name,population])
import pandas as pd
df=pd.DataFrame(all_countries,columns=(["name","population"]))
df
```
